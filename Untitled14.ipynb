```python
import chess
import numpy as np
import tensorflow as tf

# Dummy Deep Learning model (untrained, just for demo)
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(64,)),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(1)  # Score output
])

def board_to_input(board):
    """Converts the board to a 64-element vector for model input."""
    piece_map = board.piece_map()
    input_vector = np.zeros(64)
    for square in range(64):
        piece = piece_map.get(square)
        if piece:
            value = piece.piece_type
            if piece.color == chess.BLACK:
                value = -value
            input_vector[square] = value
    return input_vector

def get_best_move(board):
    """Evaluate all legal moves using the DL model and pick the best one."""
    best_move = None
    best_score = -np.inf if board.turn == chess.WHITE else np.inf

    for move in board.legal_moves:
        board.push(move)
        input_vector = board_to_input(board)
        input_vector = np.expand_dims(input_vector, axis=0)
        score = model.predict(input_vector, verbose=0)[0][0]
        board.pop()

        if board.turn == chess.WHITE and score > best_score:
            best_score = score
            best_move = move
        elif board.turn == chess.BLACK and score < best_score:
            best_score = score
            best_move = move

    return best_move

def play_game():
    board = chess.Board()

    print("Play against the AI! You are White.")
    while not board.is_game_over():
        print(board.unicode())
        if board.turn == chess.WHITE:
            move_input = input("Your move (e.g., e2e4): ").strip()
            try:
                move = chess.Move.from_uci(move_input)
                if move in board.legal_moves:
                    board.push(move)
                else:
                    print("Illegal move.")
            except:
                print("Invalid input.")
        else:
            print("AI is thinking...")
            ai_move = get_best_move(board)
            board.push(ai_move)
            print(f"AI played: {ai_move}")

    print("\nGame Over!")
    print(board.result())

if __name__ == "__main__":
    play_game()
```
